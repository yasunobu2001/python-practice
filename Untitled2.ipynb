```python
# 関数の定義

def fruit_price(number_of_momo, number_of_mikan):
    # ももとみかんの合計金額を計算する

    total_momo = number_of_momo * 100
    total_mikan = number_of_mikan * 40
    total = total_momo + total_mikan
    return total


# もも5個、みかん10個の値段を計算
total = fruit_price(5, 10)
print("もも 5個と、みかん 10個で、", total, "円です")
```

```python
def calculate_total(momo_quantity, mikan_quantity):
    momo_price = 100
    mikan_price = 40
    total_cost = (momo_quantity * momo_price) + (mikan_quantity * mikan_price)
    return total_cost


momo_quantity = int(input("購入するももの個数を入力してください: "))
mikan_quantity = int(input("購入するみかんの個数を入力してください: "))

total = calculate_total(momo_quantity, mikan_quantity)

print(f"合計金額は {total} 円です。")
```

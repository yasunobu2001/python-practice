```python
momo_quantity = int(input("購入するももの個数を入力してください: "))
mikan_quantity = int(input("購入するみかんの個数を入力してください: "))

momo_price = 100
mikan_price = 40

total_cost = (momo_quantity * momo_price) + (mikan_quantity * mikan_price)

print(f"合計金額は {total_cost} 円です。")
```
